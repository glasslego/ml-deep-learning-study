```python
"""
PyTorch 기본 연산과 자동 미분 (Autograd) 실습
========================================

학습 목표:
1. Tensor 생성 및 기본 연산
2. 자동 미분의 개념과 사용법
3. 경사하강법 구현
4. 간단한 선형 회귀 학습

실행 전 설치:
pip install torch torchvision matplotlib numpy
"""

# ============================================
# 1. 라이브러리 import
# ============================================

import torch
import numpy as np
import matplotlib.pyplot as plt

print("=" * 60)
print("PyTorch 버전:", torch.__version__)
print("CUDA 사용 가능:", torch.cuda.is_available())
print("=" * 60)


# ============================================
# 2. Tensor 기본 연산
# ============================================

print("\n" + "=" * 60)
print("섹션 1: Tensor 생성과 기본 연산")
print("=" * 60)

# 2.1 다양한 방법으로 Tensor 생성
print("\n📌 1) Tensor 생성 방법들")

# 리스트로부터 생성
tensor_from_list = torch.tensor([1, 2, 3, 4, 5])
print(f"리스트로부터: {tensor_from_list}")

# NumPy 배열로부터 생성
numpy_array = np.array([[1, 2], [3, 4]])
tensor_from_numpy = torch.from_numpy(numpy_array)
print(f"NumPy로부터:\n{tensor_from_numpy}")

# 0으로 채워진 텐서
zeros = torch.zeros(2, 3)
print(f"Zeros (2x3):\n{zeros}")

# 1로 채워진 텐서
ones = torch.ones(3, 2)
print(f"Ones (3x2):\n{ones}")

# 랜덤 텐서
random_tensor = torch.rand(2, 3)  # 0~1 균등분포
print(f"Random (2x3):\n{random_tensor}")

# 정규분포 랜덤 텐서
randn_tensor = torch.randn(2, 3)  # 평균 0, 분산 1
print(f"Randn (2x3):\n{randn_tensor}")

# 2.2 Tensor 속성
print("\n📌 2) Tensor 속성")
x = torch.randn(3, 4)
print(f"Shape: {x.shape}")
print(f"Size: {x.size()}")
print(f"Data type: {x.dtype}")
print(f"Device: {x.device}")
print(f"차원 수: {x.ndim}")

# 2.3 기본 연산
print("\n📌 3) 기본 연산")

a = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
b = torch.tensor([[5.0, 6.0], [7.0, 8.0]])

print(f"a =\n{a}")
print(f"b =\n{b}")

# 덧셈
print(f"\na + b =\n{a + b}")
print(f"torch.add(a, b) =\n{torch.add(a, b)}")

# 곱셈 (element-wise)
print(f"\na * b =\n{a * b}")

# 행렬 곱셈
print(f"\na @ b (행렬곱) =\n{a @ b}")
print(f"torch.matmul(a, b) =\n{torch.matmul(a, b)}")

# 전치
print(f"\na.T (전치) =\n{a.T}")

# 통계 함수
print(f"\na의 평균: {a.mean()}")
print(f"a의 합: {a.sum()}")
print(f"a의 최댓값: {a.max()}")
print(f"a의 최솟값: {a.min()}")

# 2.4 Shape 변경
print("\n📌 4) Shape 변경")
x = torch.randn(2, 3, 4)
print(f"원본 shape: {x.shape}")

# reshape
y = x.reshape(6, 4)
print(f"reshape(6, 4): {y.shape}")

# view (메모리 연속적일 때)
z = x.view(2, -1)  # -1은 자동 계산
print(f"view(2, -1): {z.shape}")

# squeeze (크기 1인 차원 제거)
x_with_1 = torch.randn(1, 3, 1, 4)
print(f"squeeze 전: {x_with_1.shape}")
print(f"squeeze 후: {x_with_1.squeeze().shape}")

# unsqueeze (차원 추가)
x_normal = torch.randn(3, 4)
print(f"unsqueeze 전: {x_normal.shape}")
print(f"unsqueeze(0) 후: {x_normal.unsqueeze(0).shape}")
print(f"unsqueeze(1) 후: {x_normal.unsqueeze(1).shape}")


# ============================================
# 3. 자동 미분 (Autograd) 기초
# ============================================

print("\n" + "=" * 60)
print("섹션 2: 자동 미분 (Autograd)")
print("=" * 60)

# 3.1 requires_grad 이해하기
print("\n📌 1) requires_grad - 미분 추적 활성화")

x = torch.tensor([2.0], requires_grad=True)
print(f"x = {x}")
print(f"x.requires_grad = {x.requires_grad}")

# 연산
y = x**2  # y = x^2
print(f"y = x^2 = {y}")
print(f"y.requires_grad = {y.requires_grad}")

# 역전파
y.backward()  # dy/dx 계산

print(f"dy/dx = 2x = 2*{x.item()} = {x.grad}")

# 3.2 복잡한 함수의 미분
print("\n📌 2) 복잡한 함수 미분")

# z = 3x^2 + 2x + 1의 미분
x = torch.tensor([2.0], requires_grad=True)
z = 3 * x**2 + 2 * x + 1

print(f"x = {x.item()}")
print(f"z = 3x^2 + 2x + 1 = {z.item()}")

z.backward()
print(f"dz/dx = 6x + 2 = 6*{x.item()} + 2 = {x.grad.item()}")

# 3.3 벡터 함수의 미분
print("\n📌 3) 벡터 함수의 Jacobian")

x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
y = x**2
print(f"x = {x}")
print(f"y = x^2 = {y}")

# 벡터를 스칼라로 만들기 (합 또는 평균)
z = y.sum()  # 또는 y.mean()
z.backward()

print(f"dz/dx = 2x = {x.grad}")

# 3.4 gradient를 여러 번 계산하기
print("\n📌 4) 여러 번 backward 호출")

x = torch.tensor([3.0], requires_grad=True)

for i in range(3):
    y = x**2
    y.backward()
    print(f"Step {i + 1}: x.grad = {x.grad.item()}")

    # gradient 누적 방지를 위해 초기화
    x.grad.zero_()

# 3.5 gradient 중지
print("\n📌 5) Gradient 계산 중지")

x = torch.tensor([1.0], requires_grad=True)
y = x**2

# 방법 1: detach()
z = y.detach()
print(f"z = y.detach(), z.requires_grad = {z.requires_grad}")

# 방법 2: with torch.no_grad()
with torch.no_grad():
    w = x**2
    print(f"w = x^2 (no_grad), w.requires_grad = {w.requires_grad}")


# ============================================
# 4. 경사하강법 직접 구현
# ============================================

print("\n" + "=" * 60)
print("섹션 3: 경사하강법으로 최적화")
print("=" * 60)

# 4.1 1차원 함수 최소화
print("\n📌 1) f(x) = (x-3)^2 최소화")

# 목표: f(x) = (x-3)^2의 최솟값 찾기 (정답: x=3)

x = torch.tensor([0.0], requires_grad=True)
learning_rate = 0.1
steps = 50

x_history = []
f_history = []

print(f"초기값: x = {x.item():.4f}")

for step in range(steps):
    # Forward: 함수값 계산
    f = (x - 3) ** 2

    # Backward: 미분 계산
    f.backward()

    # 기록
    x_history.append(x.item())
    f_history.append(f.item())

    # Gradient Descent 업데이트
    with torch.no_grad():
        x -= learning_rate * x.grad

    # Gradient 초기화
    x.grad.zero_()

    if (step + 1) % 10 == 0:
        print(f"Step {step + 1}: x = {x.item():.4f}, f(x) = {f.item():.4f}")

print(f"\n최종 결과: x = {x.item():.4f} (정답: 3.0)")

# 시각화
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(x_history, "b-", linewidth=2, marker="o", markersize=3)
plt.axhline(y=3, color="r", linestyle="--", label="Target (x=3)")
plt.xlabel("Step")
plt.ylabel("x")
plt.title("Optimization Path: x values")
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(f_history, "g-", linewidth=2, marker="o", markersize=3)
plt.xlabel("Step")
plt.ylabel("f(x)")
plt.title("Loss: f(x) = (x-3)²")
plt.grid(True, alpha=0.3)
plt.yscale("log")

plt.tight_layout()
plt.savefig("gradient_descent_1d.png", dpi=150, bbox_inches="tight")
plt.show()

# 4.2 2차원 함수 최소화
print("\n📌 2) f(x, y) = x^2 + y^2 최소화")

# Rosenbrock 함수의 간단한 버전
x = torch.tensor([2.0, 2.0], requires_grad=True)
learning_rate = 0.01
steps = 100

xy_history = []
loss_history = []

for step in range(steps):
    # Forward
    loss = (x[0] - 1) ** 2 + (x[1] - 1) ** 2

    # Backward
    loss.backward()

    # 기록
    xy_history.append(x.detach().clone().numpy())
    loss_history.append(loss.item())

    # Update
    with torch.no_grad():
        x -= learning_rate * x.grad

    x.grad.zero_()

    if (step + 1) % 20 == 0:
        print(
            f"Step {step + 1}: x={x[0].item():.4f}, y={x[1].item():.4f}, loss={loss.item():.4f}"
        )

# 시각화
xy_history = np.array(xy_history)

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(xy_history[:, 0], label="x", linewidth=2, marker="o", markersize=3)
plt.plot(xy_history[:, 1], label="y", linewidth=2, marker="o", markersize=3)
plt.axhline(y=1, color="r", linestyle="--", alpha=0.5, label="Target (1, 1)")
plt.xlabel("Step")
plt.ylabel("Value")
plt.title("Optimization Path: x and y")
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(loss_history, "purple", linewidth=2, marker="o", markersize=3)
plt.xlabel("Step")
plt.ylabel("Loss")
plt.title("Loss over iterations")
plt.grid(True, alpha=0.3)
plt.yscale("log")

plt.tight_layout()
plt.savefig("gradient_descent_2d.png", dpi=150, bbox_inches="tight")
plt.show()


# ============================================
# 5. 선형 회귀 구현
# ============================================

print("\n" + "=" * 60)
print("섹션 4: 선형 회귀 (Linear Regression)")
print("=" * 60)

# 5.1 데이터 생성
print("\n📌 1) 데이터 생성")

torch.manual_seed(42)

# y = 2x + 3 + noise
X = torch.randn(100, 1) * 10  # 100개 샘플, 1개 특징
true_w = 2.0
true_b = 3.0
y = true_w * X + true_b + torch.randn(100, 1) * 2  # 노이즈 추가

print(f"실제 가중치: w={true_w}, b={true_b}")
print(f"데이터 shape: X={X.shape}, y={y.shape}")

# 시각화
plt.figure(figsize=(10, 6))
plt.scatter(X.numpy(), y.numpy(), alpha=0.5, s=30, label="Data")
plt.plot(
    X.numpy(),
    (true_w * X + true_b).numpy(),
    "r-",
    linewidth=2,
    label=f"True: y={true_w}x+{true_b}",
)
plt.xlabel("X")
plt.ylabel("y")
plt.title("Generated Data for Linear Regression")
plt.legend()
plt.grid(True, alpha=0.3)
plt.savefig("linear_regression_data.png", dpi=150, bbox_inches="tight")
plt.show()

# 5.2 모델 파라미터 초기화
print("\n📌 2) 모델 학습")

# 파라미터 (학습할 값)
w = torch.randn(1, 1, requires_grad=True)
b = torch.zeros(1, 1, requires_grad=True)

print(f"초기 파라미터: w={w.item():.4f}, b={b.item():.4f}")

# 하이퍼파라미터
learning_rate = 0.01
epochs = 100

loss_history = []
w_history = []
b_history = []

# 5.3 학습 루프
for epoch in range(epochs):
    # Forward: 예측
    y_pred = X @ w + b

    # Loss: MSE (Mean Squared Error)
    loss = ((y_pred - y) ** 2).mean()

    # Backward: 미분 계산
    loss.backward()

    # Update: 경사하강법
    with torch.no_grad():
        w -= learning_rate * w.grad
        b -= learning_rate * b.grad

    # Gradient 초기화
    w.grad.zero_()
    b.grad.zero_()

    # 기록
    loss_history.append(loss.item())
    w_history.append(w.item())
    b_history.append(b.item())

    if (epoch + 1) % 20 == 0:
        print(
            f"Epoch {epoch + 1:3d} | Loss: {loss.item():.4f} | w: {w.item():.4f} | b: {b.item():.4f}"
        )

print(f"\n최종 파라미터: w={w.item():.4f}, b={b.item():.4f}")
print(f"실제 파라미터: w={true_w}, b={true_b}")
print(f"오차: w={abs(w.item() - true_w):.4f}, b={abs(b.item() - true_b):.4f}")

# 5.4 결과 시각화
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. 학습된 모델
ax = axes[0, 0]
ax.scatter(X.numpy(), y.numpy(), alpha=0.5, s=30, label="Data")
ax.plot(X.numpy(), (true_w * X + true_b).numpy(), "r-", linewidth=2, label="True")
with torch.no_grad():
    y_pred_final = X @ w + b
ax.plot(X.numpy(), y_pred_final.numpy(), "g--", linewidth=2, label="Predicted")
ax.set_xlabel("X")
ax.set_ylabel("y")
ax.set_title("Linear Regression Result")
ax.legend()
ax.grid(True, alpha=0.3)

# 2. Loss 감소
ax = axes[0, 1]
ax.plot(loss_history, "b-", linewidth=2)
ax.set_xlabel("Epoch")
ax.set_ylabel("Loss (MSE)")
ax.set_title("Training Loss")
ax.grid(True, alpha=0.3)
ax.set_yscale("log")

# 3. w 수렴
ax = axes[1, 0]
ax.plot(w_history, "g-", linewidth=2, label="Learned w")
ax.axhline(y=true_w, color="r", linestyle="--", linewidth=2, label=f"True w={true_w}")
ax.set_xlabel("Epoch")
ax.set_ylabel("w")
ax.set_title("Weight (w) Convergence")
ax.legend()
ax.grid(True, alpha=0.3)

# 4. b 수렴
ax = axes[1, 1]
ax.plot(b_history, "orange", linewidth=2, label="Learned b")
ax.axhline(y=true_b, color="r", linestyle="--", linewidth=2, label=f"True b={true_b}")
ax.set_xlabel("Epoch")
ax.set_ylabel("b")
ax.set_title("Bias (b) Convergence")
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig("linear_regression_training.png", dpi=150, bbox_inches="tight")
plt.show()


# ============================================
# 6. 종합 정리 및 퀴즈
# ============================================

print("\n" + "=" * 60)
print("📚 핵심 개념 정리")
print("=" * 60)

print("""
1️⃣ Tensor 기본
   - Tensor는 PyTorch의 핵심 데이터 구조
   - NumPy와 유사하지만 GPU 연산 지원
   - .shape, .size(), .dtype 등으로 속성 확인

2️⃣ 자동 미분 (Autograd)
   - requires_grad=True로 미분 추적 시작
   - .backward()로 역전파 수행
   - .grad에 gradient 저장됨
   - .zero_()로 gradient 초기화

3️⃣ 경사하강법
   - x_new = x_old - learning_rate * gradient
   - learning_rate가 너무 크면 발산, 작으면 느림
   - with torch.no_grad(): 로 파라미터 업데이트

4️⃣ 선형 회귀 학습 단계
   1) Forward: 예측값 계산
   2) Loss: 손실 함수 계산
   3) Backward: 미분 계산
   4) Update: 파라미터 업데이트
   5) Zero grad: gradient 초기화
""")

print("\n" + "=" * 60)
print("💡 실습 과제")
print("=" * 60)

print("""
1. f(x) = x^3 - 2x^2 + x의 최솟값을 경사하강법으로 찾아보세요
2. 학습률을 0.001, 0.01, 0.1로 바꿔가며 수렴 속도를 비교해보세요
3. 다항 회귀(y = ax^2 + bx + c)를 구현해보세요
4. 배치 크기를 조절하며 학습 속도 차이를 확인해보세요
""")
```
